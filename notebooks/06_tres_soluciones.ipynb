## 5. Por que la intermedia necesita el grado

Ninguna iteracion monotona la alcanza: es un punto fijo repulsivo. Los metodos
de orden (sub y supersoluciones, iteracion monotona) son constructivos pero
solo ven soluciones estables.

El grado de Leray-Schauder no construye nada, pero **cuenta**: como
$\deg(I-T_\lambda, B_R, 0)=1$, $i(0)=+1$ y el grado en un entorno del
intervalo de orden $[\varphi,1]$ tambien vale 1, por excision y aditividad
queda grado $1-1-1=-1\ne 0$ en el hueco, y ahi tiene que haber una tercera
solucion.

## 6. Lectura biologica

Con $\lambda = L^2$: en una reserva de largo $L > L^* = \sqrt{\lambda^*}$
coexisten dos estados estacionarios positivos. El de arriba es la poblacion
sana; el de abajo es el umbral que hay que superar para llegar a el.

Una poblacion reintroducida por debajo de esa curva se extingue aunque la
reserva sea suficientemente grande. Y si $L$ baja por debajo de $L^*$ el
pliegue desaparece de golpe: la extincion es abrupta y **no se revierte**
