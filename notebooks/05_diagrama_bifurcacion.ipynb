## 1. El mapa del tiempo

De la identidad de energia
$\tfrac12 (u')^2 + \lambda F(u) = \lambda F(\rho)$, con
$\rho = \|u\|_\infty = u(1/2)$, sale

$$T(\rho) = \sqrt{2}\int_0^{\rho} \frac{du}{\sqrt{F(\rho)-F(u)}},
\qquad \lambda = T(\rho)^2 .$$

Graficar $T$ sobre $(\theta(\alpha),\,1)$. Mostrar que diverge en los dos
extremos: en $\rho\to\theta^+$ porque $F(\rho)\to F(0)=0$ y la singularidad
en $u=0$ deja de ser integrable, y en $\rho\to 1^-$ porque $f(1)=0$ hace que
$F(1)-F(u)\sim (1-\alpha)(1-u)^2/2$ y la integral diverge logaritmicamente.

Nota numerica: el integrando tiene una singularidad **integrable** en
$u=\rho$. Comparar la cuadratura ingenua (trapecio en grilla uniforme, que da
cualquier cosa) contra la sustitucion $u=\rho-s^2$, que la cancela
exactamente.

## 3. El diagrama en forma de S

Graficar la curva $\rho \mapsto (T(\rho)^2,\ \rho)$ en el plano
$(\lambda,\ \|u\|_\infty)$, junto con la recta $\|u\|_\infty = 0$ de la
solucion trivial.

Lectura: para $\lambda<\lambda^*$ solo esta la trivial; en
$\lambda=\lambda^*$ aparece una solucion positiva degenerada; para
$\lambda>\lambda^*$ hay exactamente dos, la de abajo inestable y la de arriba
estable. Es una bifurcacion **silla-nodo**, no una bifurcacion desde la
solucion trivial: la rama positiva nace en el aire, sin tocar $u\equiv 0$. Eso
es la firma del efecto Allee fuerte, y lo que hace que la extincion sea
irreversible una vez cruzado el umbral.

## 6. El teorema de la funcion implicita

Con $\Phi(\lambda,u)=u''+\lambda f(u)$ entre
$X=\{u\in C^2: u(0)=u(1)=0\}$ e $Y=C([0,1])$, la derivada parcial
$D_u\Phi(\lambda,u)v = v'' + \lambda f'(u)v$ es isomorfismo si y solo si
$0$ no es autovalor del linealizado.

Derivando $\Phi(\lambda,u_\lambda)=0$ respecto de $\lambda$ se obtiene la
**ecuacion de la rama**

$$L_\lambda \dot u = f(u_\lambda), \qquad \dot u(0)=\dot u(1)=0 .$$

Resolverla numericamente en varios puntos de la rama y compararla con la
derivada por diferencias finitas de la rama trazada en la seccion 4. Que las
dos coincidan es la comprobacion de que la rama es efectivamente $C^1$ y de que
el TFI se esta aplicando bien.
