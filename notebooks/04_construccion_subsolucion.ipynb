## 3. Diagnostico: la delta de Dirac en la esquina

**Este es el punto central del notebook.**

En el empalme con la meseta la derivada salta *hacia abajo*, de un valor
positivo a cero. Una esquina concava introduce una delta de Dirac **negativa**
en $\varphi''$, luego una delta **positiva** en $-\varphi''$, que ninguna cota
$\lambda f(\varphi)$ puede dominar: la desigualdad
$-\varphi'' \le \lambda f(\varphi)$ se viola en sentido distribucional
exactamente en ese punto.

La regla general: se admiten esquinas **convexas** (saltos de derivada hacia
arriba, que dan deltas negativas en $-\varphi''$) y se prohiben las
**concavas**.

Ilustrarlo con un zoom sobre el empalme y, sobre todo, mostrar el modo en que
uno se equivoca: evaluar el residuo sobre una grilla uniforme que no cae
exactamente en el quiebre. La delta es invisible para esa grilla y el residuo
parece pequeno. Por eso `verify_subsolution` mira los saltos de $\varphi'$
**aparte** del residuo puntual, y no como subproducto suyo.

Segundo motivo de falla, independiente del primero: sobre la rampa
$\varphi''=0$, con lo cual $-\varphi''=0$, pero $f(\varphi)<0$ mientras
$\varphi<\alpha$; la desigualdad pide $0 \le \text{(negativo)}$ y falla en
todo el tramo inicial, no solo en un punto.
