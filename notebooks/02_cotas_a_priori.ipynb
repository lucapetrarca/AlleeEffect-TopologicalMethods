## 2. El truncamiento $\tilde f$

Graficar $f$ y su truncada $\tilde f$ sobre $[-0.5,\,1.5]$. Como
$f(0)=f(1)=0$, la truncada se anula identicamente fuera de $[0,1]$.

Ese hecho es el que hace andar el principio del maximo: donde $u<0$ la ecuacion
truncada da $-u''=0$, luego $u$ es afin y se anula en los extremos de ese
subintervalo, contradiccion. Lo mismo del otro lado con $u>1$.

Ademas $\tilde f$ es acotada y globalmente Lipschitz, mientras que $f$ es
cubica: ese cambio es lo que permite aplicar Schauder en una bola grande.

## 4. Compacidad

$T_\lambda$ es compacto en $C([0,1])$: como $\tilde f$ es acotada por una
constante $K$, la imagen de cualquier conjunto acotado esta contenida en
$\{v: \|v\|_\infty \le \lambda K/8,\ \|v'\|_\infty \le \lambda K/2\}$,
que es equicontinuo, y Arzela-Ascoli cierra el argumento.
